```python
# Simulate coin tosses
n = 10000
coin_tosses = np.random.choice(['H', 'T'], size=n)

# Estimate probability of heads
p_heads = np.mean(coin_tosses == 'H')
print(p_heads)
```

```python
# Example: dice rolls
rolls = np.random.randint(1, 7, size=10000)

A = rolls > 3        # event A
B = rolls % 2 == 0   # event B

# P(A | B)
p_A_given_B = np.sum(A & B) / np.sum(B)
print(p_A_given_B)

print(np.mean(A))
print(np.mean(B))
```
