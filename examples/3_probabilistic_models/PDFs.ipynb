```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform, norm, binom, poisson, nbinom
```

```python
np.random.seed(42)
```

```python
a, b = 0, 1
x = np.linspace(-0.5, 1.5, 100)
pdf = uniform.pdf(x, a, b-a)

samples = np.random.uniform(a, b, 1000)

plt.figure()
plt.hist(samples, bins=30, density=True, alpha=0.6, label="Samples")
plt.plot(x, pdf, label="PDF")
plt.title("Uniform Distribution")
plt.legend()
plt.show()
```

Central Limit Theorem

```python
sample_means = []
for _ in range(1000):
    sample = np.random.uniform(0, 1, 50)
    sample_means.append(np.mean(sample))

plt.figure()
plt.hist(sample_means, bins=30)
plt.title("Sample Means (n=50)")
plt.show()
```

```python
mu, sigma = 0, 1
x = np.linspace(-4, 4, 100)
pdf = norm.pdf(x, mu, sigma)

samples = np.random.normal(mu, sigma, 1000)

plt.figure()
plt.hist(samples, bins=30, density=True, alpha=0.6, label="Samples")
plt.plot(x, pdf, label="PDF")
plt.title("Gaussian Distribution")
plt.legend()
```

```python
cdf = norm.cdf(x, mu, sigma)

plt.figure()
plt.hist(samples, cumulative=True, bins=30, density=True, alpha=0.6, label="Samples")
plt.plot(x, cdf, label="CDF")
plt.title("Gaussian Distribution")
plt.legend()
```

```python
print(np.quantile(samples, 0.5))
print(np.quantile(samples, 0.9))
```

```python
n, p = 10, 0.5
k = np.arange(0, n+1)
pmf = binom.pmf(k, n, p)

samples = np.random.binomial(n, p, 1000)

plt.figure()
plt.bar(k, pmf, alpha=0.6, label="PMF")
plt.hist(samples, bins=np.arange(-0.5, n+1.5, 1), density=True, alpha=0.4, label="Samples")
plt.title("Binomial Distribution")
plt.legend()
```

```python
lam = 3
k = np.arange(0, 15)
pmf = poisson.pmf(k, lam)

samples = np.random.poisson(lam, 1000)

plt.figure()
plt.bar(k, pmf, alpha=0.6, label="PMF")
plt.hist(samples, bins=np.arange(-0.5, 15.5, 1), density=True, alpha=0.4, label="Samples")
plt.title("Poisson Distribution")
plt.legend()
```

```python
print(np.quantile(samples, 0.5))
print(np.quantile(samples, 0.9))
```

```python
n = 3
p = 0.5
k = np.arange(0, 15)
pmf = nbinom.pmf(k, n, p)

samples = np.random.negative_binomial(n, p, 1000)

plt.figure()
plt.bar(k, pmf, alpha=0.6, label="PMF")
plt.hist(samples, bins=np.arange(-0.5, 15.5, 1), density=True, alpha=0.4, label="Samples")
plt.title("Negative Binomial Distribution")
plt.legend()
```

```python
print(np.quantile(samples, 0.5))
print(np.quantile(samples, 0.9))
```
