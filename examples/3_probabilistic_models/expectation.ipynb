```python
import numpy as np
```

```python
# Discrete random variable
values = np.array([0, 1, 2])
probabilities = np.array([0.2, 0.5, 0.3])

# Expected value
expected_value = np.sum(values * probabilities)
print(expected_value)

# Variance
variance = np.sum((values - expected_value)**2 * probabilities)
print(variance)
```
