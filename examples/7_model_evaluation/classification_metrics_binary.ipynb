```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
from sklearn.preprocessing import StandardScaler

import matplotlib.pyplot as plt
import numpy as np
```

```python
dataset = load_breast_cancer(as_frame=True)
```

```python
X = dataset.data
y = dataset.target
```

```python
X
```

```python
y
```

test_size=0.95 just to get some errors:

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.95, random_state=42
)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
ml_est = LogisticRegression()
ml_est.fit(X_train, y_train)
```

```python
yhat_train = ml_est.predict(X_train)
yhat_test = ml_est.predict(X_test)
```

```python
yhat_test
```

```python
yhat_train_proba = ml_est.predict_proba(X_train)
yhat_test_proba = ml_est.predict_proba(X_test)
```

```python
yhat_test_proba
```

```python
acc_train = metrics.accuracy_score(y_train, yhat_train)
acc_test = metrics.accuracy_score(y_test, yhat_test)

print(f"accuracy train: {acc_train:.3f}")
print(f"accuracy test: {acc_test:.3f}")
```

```python
cm = metrics.confusion_matrix(yhat_test, y_test)
disp = metrics.ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=dataset.target_names)

disp.plot(cmap="Blues")
plt.title("Confusion Matrix")
plt.show()
```

```python
print(metrics.classification_report(y_test, yhat_test, target_names=dataset.target_names))
```

```python
fpr, tpr, _ = metrics.roc_curve(y_test, yhat_test_proba[:, 1])
roc_auc = metrics.auc(fpr, tpr)

plt.figure()
plt.plot(fpr, tpr, label=f"AUC = {roc_auc:.2f}")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```

```python
precision, recall, thresholds = metrics.precision_recall_curve(y_test, yhat_test_proba[:, 1])
avg_precision = metrics.average_precision_score(y_test, yhat_test_proba[:, 1])

plt.figure()
plt.plot(recall, precision, label=f"AP = {avg_precision:.2f}")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()
plt.show()

plt.figure()
plt.plot(recall, precision, label=f"AP = {avg_precision:.2f}")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()
plt.show()
```

threshold tuning:

```python
# Compute F1 score for each threshold
f1_scores = 2 * (precision * recall) / (precision + recall + 1e-8)

# Find best threshold (max F1)
best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx]

print(f"Best Threshold (F1): {best_threshold:.3f}")
print(f"Precision at best threshold: {precision[best_idx]:.3f}")
print(f"Recall at best threshold: {recall[best_idx]:.3f}")
print(f"F1 Score: {f1_scores[best_idx]:.3f}")
```

```python
# Convert probabilities → class predictions using new threshold
y_pred_opt = (yhat_test_proba[:, 1] >= best_threshold).astype(int)

print("\nClassification Report (Optimized Threshold):\n")
print(metrics.classification_report(y_test, y_pred_opt))
```
