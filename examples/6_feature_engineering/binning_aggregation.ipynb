```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
```

```python
X, y = fetch_california_housing(return_X_y=True, as_frame=True)
```

```python
X.head()
```

```python
# equal-width bins
X["HouseAge_binned"] = pd.cut(X["HouseAge"], bins=5)

X["HouseAge_custom_binned"] = pd.cut(X["HouseAge"], bins=[0, 5, 10, 20, 50, 100])

# equal-frequency bins
X["MedInc_binned"] = pd.qcut(X["MedInc"], q=3, labels=["low", "mid", "high"])

X[["HouseAge", "HouseAge_binned", "HouseAge_custom_binned","MedInc", "MedInc_binned"]].head()
```

```python
X["MedInc_binned"].value_counts()
```

```python
X["HouseAge_binned"].value_counts()
```

```python
agg = X.groupby("MedInc_binned")["AveRooms"].agg(["sum", "mean", "count"])
agg.columns = ["total_rooms", "avg_rooms", "num_houses"]

agg.head()
```

```python
X = X.merge(agg, on="MedInc_binned", how="left")

X.head()
```
