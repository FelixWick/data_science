```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.datasets import fetch_california_housing
```

```python
X, y = fetch_california_housing(return_X_y=True, as_frame=True)
```

```python
# standardization (Z-score)
scaler = StandardScaler()
X[["HouseAge_standardized", "MedInc_standardized"]] = scaler.fit_transform(X[["HouseAge", "MedInc"]])
```

```python
# normalization [0,1]
minmax = MinMaxScaler()
X[["HouseAge_normalized", "MedInc_normalized"]] = minmax.fit_transform(X[["HouseAge", "MedInc"]])
```
