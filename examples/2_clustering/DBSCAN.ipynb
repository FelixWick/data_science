```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN, KMeans, AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
```

Generate non-spherical data

```python
X, _ = make_moons(n_samples=300, noise=0.05, random_state=42)
```

IMPORTANT: scale the data

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
dbscan = DBSCAN(eps=0.3, min_samples=5)
labels = dbscan.fit_predict(X_scaled)
```

```python
plt.figure()
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels)
plt.title("DBSCAN Clustering (Moons Dataset)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

k-Means for comparison: assumes spherical clusters, fails on non-linear structures

```python
kmeans = KMeans(n_clusters=2, random_state=42).fit(X_scaled)

plt.figure()
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=kmeans.labels_)
plt.title("k-Means Clustering (Moons Dataset)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

Hierarchical clustering for comparison: not ideal for complex shapes, highly dependent on linkage choice

```python
hier = AgglomerativeClustering(n_clusters=2, linkage='ward').fit(X_scaled)

plt.figure()
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=hier.labels_)
plt.title("Hierarchical Clustering (Moons Dataset)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
