```python
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
def plot_timeseries(df, frequency=""):
    plt.figure()
    df["Close"].plot(style='r')
    plt.ylabel("stock prize", fontsize=15)
    plt.title(frequency, fontsize=15)
```

Valid intervals: [1m, 2m, 5m, 15m, 30m, 60m, 90m, 1h, 4h, 1d, 5d, 1wk, 1mo, 3mo]")

```python
df = yf.Ticker("GOOG").history(period="6mo", interval="1d")
```

```python
df
```

```python
plot_timeseries(df, "daily")
```
